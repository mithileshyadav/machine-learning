# Modify the date data into dateformat using parse_dates parameter in read_csv()

```python
df = pd.read_csv("Data/house_price_20_rows.csv", parse_dates= ["Sale_Date"])
df
```

# get the year, month, day from the date column

```python
df["Year"] = df["Sale_Date"].dt.year
df["Month"] = df["Sale_Date"].dt.month
df["Day"] = df["Sale_Date"].dt.day
df
```

# Save the csv using to_csv()

```python
df.to_csv("My_own_clean_data.csv")
```
