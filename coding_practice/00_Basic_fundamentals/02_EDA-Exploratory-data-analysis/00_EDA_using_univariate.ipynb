## Univariate Analysis => Uni = One, Variate = Related to Variable/ Related to column

# 1. Categorical data => Male/Female, True/False, Yes/No, Spam/Not Spam , 0/1

```python
df["Gender"].value_counts().plot(kind= "bar")
plt.xlabel("Gender")
plt.ylabel("Number of students")
plt.title("students Gender data")
plt.show()
```

```python
df["Gender"].value_counts().plot(kind= "pie", autopct = "%.2f")
```

```python
plt.boxplot(df["Marks"].dropna())
plt.ylabel(df["Marks"])
plt.title("Find outliers")
plt.show()
print(df["Marks"].describe())
```
