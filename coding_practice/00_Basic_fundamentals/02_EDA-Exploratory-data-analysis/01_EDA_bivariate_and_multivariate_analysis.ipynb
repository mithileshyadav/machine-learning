```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Data/student_eda_100_rows.csv")
df
```

# scatterplot(x = Numerical[x-axis] , y = Numerical[y-axis])

```python
sns.scatterplot(x=df["Study_Hours"] , y= df["Self_Study_Hours"])  ## Bi variate
plt.show
print(df["Study_Hours"].describe())
print(df["Self_Study_Hours"].describe())
```

# barplot(x = Categorical[x-axis] , y = Numerical[y-axis])

```python
sns.barplot(x = df["Gender"], y= df["Marks"],  hue = df["Age"])  ## multi variate

```

# Box plot (Numerical - Categorical)

```python
sns.boxplot(x =df["Marks"] , y = df["Gender"])
df["Marks"].describe()
```

## categorical categorical 

```python
table= pd.crosstab(df["Gender"], df["Marks_Category"])
sns.heatmap(table, annot = True, cmap = "Blues", fmt= "d")
plt.show()
```

# Pairplot for find the Outliers

```python
sns.pairplot(df, vars = ["Study_Hours", "Marks"], hue= "Gender")
plt.show()
```
